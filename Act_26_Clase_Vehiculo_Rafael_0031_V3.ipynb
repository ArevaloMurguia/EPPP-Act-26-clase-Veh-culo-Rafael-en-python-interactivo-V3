```python
# Clase Vehiculo
# Atributos: marca, modelo, año
# Funciones: leerdatos(), mostrardatos()
# Crear 2 objetos vehiculo1, vehiculo2

class Vehiculo:
    def __init__(self):
        self.marca = None
        self.modelo = None
        self.año = None

    def leerdatos(self):
        self.marca = input("Ingrese la marca: ")
        self.modelo = input("Ingrese el modelo: ")
        self.año = input("Ingrese el año: ")
        return self.marca, self.modelo, self.año

    def mostrardatos(self):
        print("Marca: ", self.marca)
        print("Modelo:", self.modelo)
        print("Año:   ", self.año)
        return self.marca, self.modelo, self.año

# Fin de la clase Vehiculo

# Crear objetos
vehiculo1 = Vehiculo()
vehiculo2 = Vehiculo()

print("Vehiculo Numero 1")
vehiculo1.leerdatos()
vehiculo1.mostrardatos()
print("-*-*-*-*-*-*-*-*-*-*-*-*")

print("Vehiculo Numero 2")
vehiculo2.leerdatos()
vehiculo2.mostrardatos()
print("-*-*-*-*-*-*-*-*-*-*-*-*")

print("Programa realizado por Arevalo Rafael NL=3")
```
